```python
price_correlation = correlation_matrix["Price"]

print(price_correlation)
```

```python
price_correlation = correlation_matrix["Price"].drop("Price")

print(price_correlation)
```

```python
top_4_features = price_correlation.abs().sort_values(ascending=False).head(4)

print("Top 4 features most correlated with Price:")
print(top_4_features)
```

```python
top_4_feature_names = top_4_features.index.tolist()

print("Top 4 features:")
for feature in top_4_feature_names:
    print(feature)
```

```python
print("Top 4 Features and their correlation with Price:\n")

for feature in top_4_feature_names:
    correlation = price_correlation[feature]
    print(f"{feature}: {correlation:.4f}")
```

```python
for feature in top_4_feature_names:

    plt.figure(figsize=(7, 5))

    sns.scatterplot(
        data=df,
        x=feature,
        y="Price")

    plt.title(f"{feature} vs Price")
    plt.xlabel(feature)
    plt.ylabel("Price")

    plt.show()
```

RAM vs Price:Higher RAM is generally associated with higher mobile phone prices.
Internal Memory vs Price:Mobile phones with higher internal memory generally have higher prices.
PPI vs Price:Higher screen pixel density (PPI) is generally associated with higher mobile phone prices.
Rear Camera vs Price:Mobile phones with better rear-camera specifications generally tend to have higher prices.


```python
for feature in top_4_feature_names:

    plt.figure(figsize=(7, 5))

    sns.regplot(
        data=df,
        x=feature,
        y="Price"
    )

    plt.title(f"{feature} vs Price")
    plt.xlabel(feature)
    plt.ylabel("Price")

    plt.show()
```

RAM vs Price: The regression line shows a strong positive relationship, indicating that price generally increases as RAM increases.
Internal Memory vs Price: The regression line shows a positive relationship, with higher internal memory generally associated with higher prices.
PPI vs Price: The regression line indicates a strong positive relationship, where higher PPI generally corresponds to higher prices.
Rear Camera vs Price: The regression line shows a positive relationship, indicating that better rear-camera specifications generally correspond to higher prices

##2. Prepare the Data
2.1 Feature Selection

```python
# Select features and target
X = df[top_4_feature_names]
y = df["Price"]

print("Features:")
print(X.head())

print("\nTarget:")
print(y.head())
```

##2.2 Split the Dataset

```python
from sklearn.model_selection import train_test_split

# Split the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42
)

print("Training features shape:", X_train.shape)
print("Testing features shape:", X_test.shape)
print("Training target shape:", y_train.shape)
print("Testing target shape:", y_test.shape)
```

##3. Build and Train the Model:
3.1. Create a Linear Regression Model: Build a linear regression model using the training data.
3.2. Train the Model: Fit the model to the training data.

```python
from sklearn.linear_model import LinearRegression

# Create the Linear Regression model
model = LinearRegression()

print("Linear Regression model created successfully.")
```

```python
# Train the model using the training data
model.fit(X_train, y_train)

print("Model trained successfully.")
```

##4. Evaluate the Model:
4.1. Predict: Use the model to make predictions on the test set.
4.2. Metrics Calculation: Evaluate the model’s performance using the following metrics:
❖ Slope (Coefficient) and Intercept: Print the slope (coefficient) and intercept of the regression line.
❖ Model Performance Metrics: Calculate and report the following metrics:
❖ R² Score: How well does the model explain the variance in the target variable?
❖ Mean Absolute Error (MAE): What is the average absolute difference between predicted and actual values?
❖ Mean Squared Error (MSE): What is the average of the squared differences between predicted and actual values?

```python
# 4.1 Predict the target values for the test set

y_pred = model.predict(X_test)

print("Predicted values:")
print(y_pred)
```

```python
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error

# Slope (Coefficient)
slope = model.coef_[0]

# Intercept
intercept = model.intercept_

# R² Score
r2 = r2_score(y_test, y_pred)

# Mean Absolute Error
mae = mean_absolute_error(y_test, y_pred)

# Mean Squared Error
mse = mean_squared_error(y_test, y_pred)

# Display results
print("----- Model Evaluation -----")
print("Slope (Coefficient):", slope)
print("Intercept:", intercept)
print("R² Score:", r2)
print("Mean Absolute Error (MAE):", mae)
print("Mean Squared Error (MSE):", mse)
```
